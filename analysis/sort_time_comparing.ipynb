```python
import pandas as pd

data = pd.read_csv('sorting_performance.csv')
```

```python
comparators = data['Comparator'].unique()
sizes = data['Table Size'].unique()
algorithms = data['Algorithm'].unique()
times_fastest = {algorithm: 0 for algorithm in algorithms}
sizes_insertion_isnt_fastest = []
for comparator in comparators:
    subset = data[data['Comparator'] == comparator]
    for size in sizes:
        size_subset = subset[subset['Table Size'] == size]
        #One entry per algorithm, so we can just take the first one for each algorithm
        times = []
        for algorithm in algorithms:
            alg_subset = size_subset[size_subset['Algorithm'] == algorithm]
            times.append((algorithm, int(alg_subset.iloc[0]['Average Time (ns)'])))
        times = sorted(times, key=lambda x: x[1]) 
        times_fastest[times[0][0]] += 1
        if( times[0][0] != "insertion" ):
            print(times)
            sizes_insertion_isnt_fastest.append(int(size))
            
print(times_fastest)
print("Sizes where insertion isn't fastest:", sorted(list(set(sizes_insertion_isnt_fastest))))

```
