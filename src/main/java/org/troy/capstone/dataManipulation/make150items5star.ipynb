```python
import org.troy.capstone.utils.TableUtils;
import tech.tablesaw.api.Table;
import tech.tablesaw.api.FloatColumn;
import org.troy.capstone.constants.tableColumns;
import java.util.Random;
import java.util.Collections;
import java.util.List;
import java.util.ArrayList;
import java.util.stream.IntStream;
import java.util.stream.Collectors;
import java.util.function.DoublePredicate;

//Load the data and get column
Table table = TableUtils.readCleanedAttributedDataLongPath();
FloatColumn ratingColumn = table.floatColumn(tableColumns.REVIEW_SCORE.getColumnName());

// Count current 5-star items
int numberOf5StarItems = ratingColumn.count((DoublePredicate) d -> d == 5.0);
System.out.println("Current 5-star items: " + numberOf5StarItems);

List<Integer> rowIndices = IntStream.range(0, table.rowCount()).boxed().collect(Collectors.toList());

// Shuffle the indices randomly
Collections.shuffle(rowIndices);

int itemsToUpdate = 150 - numberOf5StarItems;
if (itemsToUpdate > 0) {
    System.out.println("Adding " + itemsToUpdate + " more 5-star items");
    for (int i = 0; i < itemsToUpdate; i++) {
        ratingColumn.set(rowIndices.get(i).intValue(), 5.0f);
    }
} else {
    System.out.println("Already have " + numberOf5StarItems + " 5-star items (target was 150)");
}

TableUtils.writeCleanedAttributedDataLongPath(table);

// Count final 5-star items
int finalCount = ratingColumn.count((DoublePredicate) d -> d == 5.0);
System.out.println("Final 5-star items count: " + finalCount);
```
