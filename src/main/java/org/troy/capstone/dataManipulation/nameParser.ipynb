```python
import tech.tablesaw.api.Table;
import org.troy.capstone.constants.tableColumns;
import org.troy.capstone.utils.TableUtils;
import tech.tablesaw.api.StringColumn;
import org.troy.capstone.constants.dataPaths;

Table table = TableUtils.readCleanedDataLongPath();
StringColumn nameColumn = table.stringColumn(tableColumns.NAME.getColumnName());
for( int x = 0; x < table.rowCount(); x++) {
    String name = nameColumn.get(x).trim();
    String[] splitName = name.split(" ");
    if( splitName[splitName.length - 1].matches("\\d+") )
        splitName = Arrays.copyOf(splitName, splitName.length - 1);
    String query = String.join(" ", splitName);
    nameColumn.set(x, query);
}

TableUtils.writeCleanedDataLongPath(table);
```
