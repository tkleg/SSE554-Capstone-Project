```python
import tech.tablesaw.api.Table;
import org.troy.capstone.constants.tableColumns;
import org.troy.capstone.utils.TableUtils;
import tech.tablesaw.api.StringColumn;
import org.troy.capstone.constants.dataPaths;

Table table = TableUtils.readCleanedAttributedDataLongPath();
StringColumn photoAuthorColumn = table.stringColumn(tableColumns.PHOTO_AUTHOR.getColumnName());
for( int row = 0; row < table.rowCount(); row++) {
    String input = photoAuthorColumn.get(row);
    String[] words = input.toLowerCase().split("\\s+");
    StringBuilder result = new StringBuilder();

    for (String word : words)
        result.append(Character.toUpperCase(word.charAt(0)))
              .append(word.substring(1))
              .append(" ");

    photoAuthorColumn.set(row, result.toString().trim());
}
TableUtils.writeCleanedAttributedDataLongPath(table);
```
