## 2. Variables and Basic Operations 🔢

Java supports various data types and operations:

```python
// Declare variables of different types
int number = 42;
double pi = 3.14159;
String message = "Java in Jupyter is awesome!";
boolean isWorking = true;

// Perform operations
int result = number * 2;
double circleArea = pi * 5 * 5; // radius = 5

// Print results
System.out.println("Integer: " + number);
System.out.println("Double: " + pi);
System.out.println("String: " + message);
System.out.println("Boolean: " + isWorking);
System.out.println("Calculation result: " + result);
System.out.println("Circle area (radius=5): " + circleArea);
```

## 3. Arrays and Collections 📚

Working with arrays and lists in Java:

```python
import java.util.*;

// Create and work with arrays
int[] numbers = {1, 2, 3, 4, 5};
String[] languages = {"Java", "Python", "JavaScript"};

// Create and work with ArrayList
List<String> fruits = new ArrayList<>();
fruits.add("Apple");
fruits.add("Banana");
fruits.add("Orange");

// Print array contents
System.out.println("Numbers array: " + Arrays.toString(numbers));
System.out.println("Languages array: " + Arrays.toString(languages));

// Print list contents
System.out.println("Fruits list: " + fruits);
System.out.println("First fruit: " + fruits.get(0));
System.out.println("List size: " + fruits.size());
```

## 4. Classes and Methods 🏗️

Creating and using classes in Java:

```python
// Define a simple Person class
class Person {
    private String name;
    private int age;
    
    public Person(String name, int age) {
        this.name = name;
        this.age = age;
    }
    
    public String getName() {
        return name;
    }
    
    public int getAge() {
        return age;
    }
    
    public String introduce() {
        return "Hi, I'm " + name + " and I'm " + age + " years old.";
    }
}

// Create and use Person objects
Person person1 = new Person("Alice", 25);
Person person2 = new Person("Bob", 30);

System.out.println(person1.introduce());
System.out.println(person2.introduce());
System.out.println("Alice's age: " + person1.getAge());
System.out.println("Bob's name: " + person2.getName());
```
